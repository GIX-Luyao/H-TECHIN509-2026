## 2. String methods — cleaning messy text

Real people type messy text: `"   How MANY vacation  days do I GET?  "`. Before your assistant
uses a question, it tidies it. A **string method** is a cleanup move you call with a dot.

One at a time:

````python
raw = "   How MANY vacation days do I GET?  "

print(raw)                  # shows the string as-is, with spaces
print(repr(raw))            # repr() shows the string with quotes so hidden spaces are visible
print(repr(raw.strip()))    # .strip()  removes spaces at the ends
print(repr(raw.lower()))    # .lower()  makes everything lowercase
````

````python
# Two more you may use:
print(raw.replace("vacation", "PTO"))    # .replace(old, new) swaps text
print(raw.split())                        # .split() breaks a string into a list of words
````

**Note:** string methods do **not** change the original. They hand
you a **new** string. Run the next cell and look carefully:

````python
raw.strip()          # this made a cleaned copy…
print(repr(raw))     # …but raw is UNCHANGED — the copy was thrown away!

cleaned = raw.strip().lower()   # the fix: SAVE the result in a variable
print(repr(cleaned))            # (chaining: strip first, then lower the result)
````

## 4. f-strings

You have variables. You have clean text. Now the move that assembles them into sentences:
an **f-string**. Put `f` before the opening quote, then write `{variable}` inside the text —
Python swaps in the value.

````python
greeting = f"You are {my_persona}. Speak in a {my_tone} way."
print(greeting)
````

### A classic mistake--Forgetting `f` 

Forget the `f` and Python doesn't fill in the blanks

````python
oops = "You are {my_persona}. Speak in a {my_tone} way."    # <- no f!
print("without f:", oops)

fixed = f"You are {my_persona}. Speak in a {my_tone} way."   # <- f restored
print("with f:   ", fixed)
````

> **You may also see `.format()`** in older code:
> `"You are {}.".format(my_persona)`. f-strings are preferred.

## What is on the other end of the wire?

Please note:

**Your text goes in as pieces, not as sentences.** The model does not read your prompt the way you
just did. It is first chopped into **tokens** — word-ish pieces, roughly four characters each. Common
words survive whole; rarer ones split apart.

Two things follow, and both come back later:

- **You are billed per token**, not per sentence. Week 10 builds the cost model on this.
- **There is a hard limit on how many tokens the model can look at in one request.** That limit is
  the **context window**, and it explains two things you will otherwise hit blind: why the `messages`
  list from Week 3 cannot grow forever, and why Week 6 chops a long document into chunks instead of
  pasting the whole handbook into the prompt.

**Why your paragraph works at all.** The model was trained in two stages:

| Stage | What happened | What you get |
|---|---|---|
| **Pretraining** | Given the tokens so far, predict the next one — over an enormous amount of text | A **text continuer**. Apparent knowledge, but no instinct to obey you |
| **Instruction tuning** | Trained further on examples of instructions being *followed* | Something that treats your paragraph as a standing order |

That second stage is the reason this week is possible. Hand `You are a helpful HR assistant. Always
cite the policy section.` to a pretrained-only model and it is liable to write you *more job
description* — because that is what text like that is usually followed by.

So: **the system prompt is not a setting and it is not configuration. It is just more tokens, placed
in front of yours, read by a model trained to treat the first ones as instructions.** That is the
whole mechanism.

> Hold onto that sentence until **Week 8**. If instructions are only text, then text sitting inside a
> document you retrieved can read as instructions too. That is prompt injection, and it is a direct
> consequence of what you just learned — not a separate mystery.

````python
# Tokens, and the wall your conversation eventually hits.
# The split below is what a real tokenizer produces for this sentence — shown rather than
# computed, because we don't need the tokenizer library to see the idea (Week 10 estimates them).

question = "how many vacation days do i get?"
tokens = ["how", " many", " vac", "ation", " days", " do", " i", " get", "?"]

print(f"the sentence : {question!r}")
print(f"words        : {len(question.split())}")
print(f"tokens       : {len(tokens)}   <- more pieces than words")
print(f"the pieces   : {tokens}")
print()
print('note "vacation" came apart into " vac" + "ation" — common words survive, rarer ones split')

# The context window: a hard ceiling on tokens per request.
CONTEXT_WINDOW = 8_000        # a small model's budget, for the sake of arithmetic
tokens_per_turn = 60          # a short question plus a short answer

print()
print(f"context window  : {CONTEXT_WINDOW:,} tokens")
print(f"cost per turn   : ~{tokens_per_turn} tokens")
print(f"turns until full: ~{CONTEXT_WINDOW // tokens_per_turn}")
print()
print("That ceiling is why Week 3's messages list can't grow forever,")
print("and why Week 6 chunks documents instead of pasting whole files.")
````

### The knob: `temperature`

At every step the model produces a probability for *every* possible next token. **Temperature**
decides how far it is allowed to wander from the most likely one:

```py
from ribot.llm import chat

chat(messages)                      # temperature=0.2 — the course default
chat(messages, temperature=0.0)     # near-deterministic: same answer each run
chat(messages, temperature=1.0)     # more varied, less predictable
```

Low temperature is what you want whenever an answer should be reproducible — which is why `ribot`
ships at `0.2`, and why the golden-set measurements you take in Week 6 are worth trusting at all. A
number that changes every time you run it cannot be evidence of anything.

> **Offline note:** the `FakeLLM` you are running this week is deterministic and ignores
> `temperature` entirely — there is no sampling to adjust. You will see the knob do real work after
> Key Day in Week 5.

## 8. A good practice: predict before you run

You just changed one variable and watched the whole prompt change with it. That was an
**experiment**

Why bother? Because *"this prompt works better"*, *"semantic search is smarter"*, and
*"the local model is good enough"* are **claims**. The people who are genuinely useful around
AI systems are the ones who can turn a claim into a number — or at least into a recorded
observation with a prediction attached. 

**The five lines.** 

| Line | What goes in it |
|---|---|
| **Hypothesis** | What you believe, stated so it *could* be proven wrong |
| **Change** | The ONE thing you change |
| **Held fixed** | What you deliberately leave alone |
| **Metric & prediction** | What you'll look at — and your guess, **written before you run** |
| **Result & decision** | What actually happened, and what you'll do about it |

Each line enforces a rule that gets broken constantly in real projects: predict before you
look, change one thing at a time, keep a baseline to compare against, name your metric up
front, and end with a decision.
