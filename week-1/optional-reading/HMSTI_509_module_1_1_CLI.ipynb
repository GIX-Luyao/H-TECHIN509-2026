# Command line tools, Shells, Terminal Emulator
Computer systems tend to have a:
* **GUI (graphical user interface)**: Users interact with computers using visual elements such as windows, icons, and menus. Think of Windows, Linux desktop, MacOS GUI.
* **CLI (command line interface)**: Users interact with computers by typing commands into a terminal, which are interpreted by a shell. 
* These are both interface to the operating system (Windows has Windows Shell, the old DOS had command.com program as a shell, Mac OS X is Unix-based operating system usually has Z shell or Bash)
* In applications, like a website or mobile app, there is usually a GUI as well as an API for command line like interaction (for automation and machine access)

# Command line programming in a Shell

Command-line programming in a shell uses a text-based interface to interact with the operating system by entering commands. 

* prompt (usually your login name, sometimes the current path or directory, and an indicator like '\$', '#', or '>')
* command (a command that the operating system understands in the shell language, like `dir`, `ls`, `echo`)
* optional argument (`$ls ~` is listing files in the user home directory, aliased with the tilde ~)
* output (usually occurs on the next line)


## Common shell commands

* `cd`: change directory
* `ls`: list files and directories
* `mkdir`: create a directory
* `pwd`: see the current directory path
* `cat`: concatenate files
* `more`: print out a file
* `man`: see manual for a command
* `>`: redirect the output of a command into a file
* `|`: pipe operator to send the output of a command into another command
* `echo`: write arguments to standard output

# Setup Virtual Environment

A **virtual envrionment** is a self-contained location for your Python project.

Python virtual environment allows us to decouple and separate installs and packages for different projects. This leads to the following advantages:
- Ensure correct packages and versions are installed for specific projects.
- Make results from codes reproducible
- Avoid potential conflict among packages or Python versions.


There are different ways to create virtual environments. We will use `venv` from Python. More details can be found here: https://docs.python.org/3/library/venv.html

Follow the steps below to create a virtual environment.
1. Assume we will store the project in a directory on the Desktop. We first navigate to the Desktop and create the directory using `mkdir`.
2. We `cd` into the folder. 
3. Create a virtual environment. Let `<env_name>` be the name to your virtual environment, then
   - if you are a Windows user, type `python -m venv <env_name>`
   - if you are a MacOS user, type `python3 -m venv <env_name>`
4. Activate the environment. 
   - if you are a Windows user, type `<env_name>/Scripts/Activate.ps1`
   - if you are a MacOS user, type `source <env_name>/bin/activate`
5. You now should see `<env_name>` appear as a prefix in your terminal.
6. Let's try to install some packages. The package to be installed should be determined by your project. Let's first see what are the packages available now using `pip list`. Here we use requests as an example to see how to install packages in virtual environment.
7. After installing the packages, we can also document all packages into a txt file so that others can easily configure their virtual environment when using your project. The command is `pip freeze > requirements.txt`. You should now see the file in the directory.
8. If you would like to use the requirements.txt file from others, you can simply use `pip install -r requirements.txt`
9. We can deactivate the virtual environment by typing `deactivate`.
